```python
# 10 profit/loss product
df["total_cost_lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    + df["Salary_Cost_Lakh"]
    +df["Other_Cost_Lakh"]
)

df["profit_lakh"] = df["Revenue_Lakh"] - df["total_cost_lakh"]
print(df[["Product" , "Revenue_Lakh" , "total_cost_lakh" , "profit_lakh"]].head(10))
```

```python
#missing value
print(df.isna().sum())
```

```python
#decision making structures
#if statement
rating = df["Customer_Rating"].mean()

if rating >=4:
    print("customer satisfaction is good")
```

```python
# decision making structures
#if else
revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("average revenue is high")
else:
  print("average revenue is low")
```

```python
import pandas as pd

# 1. Load the dataset since the runtime was reset
try:
    df = pd.read_excel("/content/Lecture2_data-2.xlsx")
except FileNotFoundError:
    print("Error: Could not find '/content/Lecture2_data-2.xlsx'. Please verify the file path.")

# 2. Check if df is successfully defined, then calculate the average rating
if 'df' in locals() or 'df' in globals():
    # Use the correct column name 'Customer_Rating'
    rating = df["Customer_Rating"].mean()
    print(f"Average Customer Rating: {rating:.2f}")

    # make a decision based on the rating
    if rating >= 4.5:
      print("excellent")
    elif rating >= 4.0:
      print("good")
    elif rating >= 3.0:
      print("average")
    else:
      print("need improvement")
```

```python
# for loop
for rating in df["Customer_Rating"]:

  if rating >= 4:
    print("good rating")
else:
  print("needs improvement")
```

```python
# for loop with range () iloc= integer location

for i in range(10):
  print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#while loop in python
#while loop and i,j is variable to track
i=0

while i < 5:
  revenue = df["Revenue_Lakh"].iloc[i]
  if revenue >= 300:
    print("revenue is high")
  else:
    print("revenue is low")
    print(revenue)
  i = i + 1
```
